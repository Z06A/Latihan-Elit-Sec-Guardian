*2.Python Scripting:*

Berikut adalah contoh skrip Python yang menggunakan pustaka Scapy untuk melakukan operasi packet sniffing dasar:

```python
#!/usr/bin/env python
 
from scapy.all import *
 
def packet_handler(packet):
    if packet.haslayer(TCP):
        print(packet[TCP].payload)
 
sniff(filter="tcp", prn=packet_handler)
```

Skrip ini menggunakan pustaka Scapy untuk mengendus paket TCP dan mencetak muatan setiap paket yang diambil.

- *2.Pemindaian Port Tingkat Lanjut dengan PowerShell:*

```python
$target = "192.168.1.1-100"
$results = 1..100 | ForEach-Object {Test-NetConnection -ComputerName $target -Port $_ -InformationLevel "Quiet"} | Where-Object {$_.TcpTestSucceeded}
Write-Output $results | Format-Table -AutoSize
```

Perintah PowerShell ini melakukan pemindaian port pada berbagai alamat IP dan menampilkan port yang terbuka.
