**Contoh Kode Tingkat Lanjut (Python):**

```python
import socket
import struct
 
def send_packet(dest_ip, dest_port, payload):
    # Buat soket mentah
    s = socket.socket(socket.AF_INET, socket.SOCK_RAW, socket.IPPROTO_TCP)
 
    # mengatur header IP secara menual
    packet = struct.pack('!BBHHHBBH4s4s', 69, 0, 40, 12345, 0, 255, 6, 0, socket.inet_aton('127.0.0.1'), socket.inet_aton(dest_ip))
 
    # mengatur header TCP secara menual
    tcp_header = struct.pack('!HHLLBBHHH', 12345, dest_port, 0, 0, 80, 0, 0, 0, 0)
 
    # membangun paket
    packet += tcp_header + payload.encode()
 
    # kirim paket
    s.sendto(packet, (dest_ip, 0))
 
# contoh penggunaan
send_packet('192.168.0.1', 8080, 'GET / HTTP/1.1\r\nHost: example.com\r\n\r\n')
```

Kode Python ini menunjukkan cara membuat paket TCP / IP mentah dan mengirimkannya ke alamat IP dan port tujuan yang ditentukan. Ini secara manual membangun header IP dan TCP menggunakan modul, memungkinkan Anda untuk menyesuaikan berbagai bidang.

**Berikut adalah contoh skrip yang menggabungkan penggunaan alat Kali Linux dengan skrip Bash untuk tujuan keamanan jaringan:**

```python
#!/bin/bash
 
# install alat yang di perlukan jika belum di insatall
apt-get update
apt-get install -y nmap wireshark snort

# lakukan pemindaian jaringan menggunakan Nmap
nmap -p 1-1000 <target_IP>
 
# tangkap lalulintas jaringan menggunakan Wireshark
wireshark -i eth0 &
 
# siapkan aturan snort untuk deteksi instrusi
echo 'alert icmp any any -> $HOME_NET any (msg:"ICMP detected"; sid:1000001; rev:1;)' > /etc/snort/rules/local.rules
service snort restart
```

Skrip ini dimulai dengan memastikan bahwa alat yang diperlukan (Nmap, Wireshark, dan Snort) diinstal pada sistem. Kemudian melanjutkan untuk melakukan pemindaian jaringan menggunakan Nmap untuk memindai port pada IP target yang ditentukan. Selanjutnya, ia meluncurkan Wireshark untuk menangkap lalu lintas jaringan pada antarmuka. Akhirnya, ia menyiapkan aturan Snort dasar untuk mendeteksi aktivitas ICMP dan memulai ulang layanan Snort.
