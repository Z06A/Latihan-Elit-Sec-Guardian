### 1.Bash Scripting:

- Bash adalah bahasa scripting yang kuat yang biasa digunakan di lingkungan Linux. 

***Berikut adalah contoh aturan firewall dasar menggunakan iptables dalam skrip Bash:***

```python
#!/bin/bash
 
# Siram aturan ayng ada

iptables -F
 
# Tetapkan Kebijakan Default

iptables -P INPUT DROP
iptables -P FORWARD DROP
iptables -P OUTPUT ACCEPT
 
# Izinkan Lalulintas Ssh mauk dari alamat ip tertentu

iptables -A INPUT -p tcp --dport 22 -s 192.168.0.100 -j ACCEPT
 
# Izinkan Lalulintas Http masuk

iptables -A INPUT -p tcp --dport 80 -j ACCEPT
 
# Izinkan lalu lintas dns keluar

iptables -A OUTPUT -p udp --dport 53 -j ACCEPT
 
# Tilak Lalulintas masuk dan keluar lainnya

iptables -A INPUT -j DROP
iptables -A OUTPUT -j DROP
```

Skrip ini menyiapkan konfigurasi firewall dasar yang memungkinkan lalu lintas SSH dan HTTP masuk sementara hanya mengizinkan lalu lintas DNS keluar. Semua lalu lintas masuk dan keluar lainnya diblokir.

### 2.Skrip Python:

- Python adalah bahasa scripting serbaguna dengan banyak pustaka dan modul untuk tugas jaringan dan keamanan siber. 

**Berikut adalah contoh skrip Python yang menggunakan pustaka untuk manipulasi paket:**
- scapy

```python
#!/usr/bin/env python
 
from scapy.all import *
 
# Tentukan aturan firewall dasar untuk memblokir paket yanng masuk

def block_packets(packet):
    if IP in packet:
        packet.drop()
 
# Mengendus paket dan menerapkan aturan firewall

sniff(filter="ip", prn=block_packets)
```

Skrip ini menggunakan pustaka untuk mengendus paket masuk dan menerapkan aturan firewall dasar untuk menjatuhkan semua paket IP masuk.scapy

*Skrip Bash Ultra-Canggih:*

```python
#!/bin/bash
 
# ini adalah skrip bash python ultra canggih untung keamanan jaringan 
 
# tentukan fungsi kostum untuk konfigurasi firewall tingkat lanjut

# berfungsi untuk memlokir lalu lintas masuk dari negara tertentu

block_country() {
    local country_code=$1
    iptables -A INPUT -m geoip --src-cc "$country_code" -j DROP
}
 
# Berfungsi untuk mencatat koneksi keluar yang mencurigrakan

log_outgoing_connections() {
    tcpdump -i eth0 -s 0 -w outgoing_connections.pcap host not 192.168.0.0/16
}
 
# Eksekusi Skrip utama
 
# Siram Aturan Yang ada
iptables -F
 
# Tetapkan Kebijakan Default
iptables -P INPUT DROP
iptables -P FORWARD DROP
iptables -P OUTPUT ACCEPT
 
# Izinkan Lalulintas ssh masuk dari alamat ip tertentu

iptables -A INPUT -p tcp --dport 22 -s 192.168.0.100 -j ACCEPT
 
# Blokir lalulintas ssh masuk dari alamat ip terttentu

block_country "CN"  # China
block_country "RU"  # Russia
 
# log Koneksi keluar

log_outgoing_connections
```

Dalam skrip Bash ultra-canggih ini, kami telah menambahkan dua fungsi kustom: untuk memblokir lalu lintas masuk dari negara tertentu menggunakan modul, dan untuk mencatat koneksi keluar yang mencurigakan menggunakan . Fungsi-fungsi ini menunjukkan konfigurasi firewall canggih dan teknik pemantauan jaringan.block_country,geoip,log_outgoing_connections,tcpdump

*Skrip Python Ultra-Canggih:*

```python
#!/usr/bin/env python
 
from scapy.all import *
import os
 
# ini adalah Skript keamanan Ultra-canggihh untuk keamanan jaringan
 
# Berfungsi unntuk melakukan inpeksi dan penyarinngan paket lanjutan

def advanced_packet_inspection(packet):
    if IP in packet:
        # Blokir paket dari alamat ip tertentu
        if packet[IP].src == "192.168.0.100":
            return
 
        # mendeteksi dan mencatata potensi lalulintas berbahaya
        if packet[IP].payload:
            payload = packet[IP].payload.original
            if "malicious_keyword" in payload:
                with open("malicious_traffic.log", "a") as log_file:
                    log_file.write(str(packet))
                    log_file.write("\n")
 
        # lakukan inpeksi dan tindakan paket lanjutan tambahan
 
# eksekusi skript utama
 
# siram aturan iptables yang ada
os.system("iptables -F")
 
# tetapkan kebijakan default
os.system("iptables -P INPUT DROP")
os.system("iptables -P FORWARD DROP")
os.system("iptables -P OUTPUT ACCEPT")
 
# Sniff paket dan menerapkan inpeksi paket jaringan

sniff(filter="ip", prn=advanced_packet_inspection)
```

Dalam skrip Python ultra-canggih ini, kami telah menambahkan kemampuan inspeksi dan pemfilteran paket tingkat lanjut. Fungsi ini menunjukkan cara memblokir paket dari alamat IP tertentu, mendeteksi dan mencatat potensi lalu lintas berbahaya, dan melakukan inspeksi dan tindakan paket lanjutan tambahan.advanced_packet_inspection
