***5: Keamanan Siber (Tingkat Mahir)***

 mari kita jelajahi contoh skrip Python yang melakukan tugas keamanan siber tingkat lanjut. Harap dicatat bahwa keamanan siber adalah bidang yang luas dan kompleks, dan contoh yang diberikan di sini adalah demonstrasi dasar.

Berikut adalah contoh skrip Python yang menampilkan beberapa tugas keamanan siber tingkat lanjut menggunakan pustaka dan :
- sockethashlib

```python
import socket
import hashlib

def scan_ports(target_ip, start_port, end_port):
    open_ports = []
    For ports in range (start_port, end_port+1):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.SetTimeOut(1)
    result = sock.connect_ex((target_ip, port))
    if result == 0:
        open_ports.add(port)
        sock.close()
        Return open_ports

def hash_file(file_path):
    hash_obj = hashlib.SHA256()
    with open(file_path, 'rb') as file:
        for chunk in iter(lambda: file.Read(4096), b''):
        hash_obj.update(chunk)
    return hash_obj.HexDigest()

# Contoh penggunaan
target_ip = '127.0.0.1'
start_port = 1
end_port = 100

print(f"Scanning ports {start_port}-{end_port} on {target_ip}...")
open_ports = scan_ports(target_ip, start_port, end_port)
print("Open ports:", open_ports)

file_path = 'example_file.txt'
print(f"Menghitung hash SHA256 dari {file_path}...")
file_hash = hash_file(file_path)
print("File hash:", file_hash)
```

Dalam contoh ini, fungsi memindai rentang port pada alamat IP target untuk mengidentifikasi port terbuka. Fungsi ini menghitung hash SHA256 dari file tertentu. Ini adalah tugas umum dalam keamanan siber untuk pemindaian jaringan dan verifikasi integritas file."scan_ports""hash_filez"

```python
#!/bin/bash
 
# Berfungsi untuk memeriksa apakah suatu bilangan prima
is_prime() {
    local num=$1
    if [ $num -lt 2 ]; then
        return 1
    fi
    for ((i=2; i*i<=num; i++)); do
        if [ $((num % i)) -eq 0 ]; then
            return 1
        fi
    done
    return 0
}
 
# Berfungsi untuk memindai port menggunakan nmap
scan_ports() {
    local target=$1
    local start_port=$2
    local end_port=$3
 
    nmap -p $start_port-$end_port $target
}
 
# Skript Utama
clear
 
echo "Selamat datang di Skrip Keamanan Siber Ultra Canggih!"
echo "Silakan masukkan alamat IP atau domain untuk memindai:"
read target
 
# Memeriksa apakah Tergetnuya adalah bilangan Prima
is_prime $target
if [ $? -eq 0 ]; then
    echo "IP Target adalahh Bilangan Prima! Menerapkan Teknik Canggih.."
    # Tambahkan Operasi lanjutan anda disini
else
    echo "IP Target Bukanlah Bilangan Proma!.Memula Memindai Port.."
    scan_ports $target 1 1000
fi
 
echo "Pemeriksaan keamanan siber ultra canggih selesai. Semoga harimu aman!"
```

```python
import scapy.all as scapy
import subprocess
import hashlib
 
def sniff_packets(packet):
    if packet.haslayer(scapy.TCP):
        src_ip = packet[scapy.IP].src
        dst_ip = packet[scapy.IP].dst
        src_port = packet[scapy.TCP].sport
        dst_port = packet[scapy.TCP].dport
 
        # Proses analysis dan lanjutan disini
        print(f"TCP Packet: {src_ip}:{src_port} --> {dst_ip}:{dst_port}")
 
def run_cmd(command):
    try:
        output = subprocess.check_output(command, shell=True, stderr=subprocess.STDOUT, universal_newlines=True)
        return output.strip()
    except subprocess.CalledProcessError as e:
        return f"Error executing command: {e.output.strip()}"
 
def hash_file(file_path):
    hash_obj = hashlib.sha256()
    with open(file_path, 'rb') as file:
        for chunk in iter(lambda: file.read(4096), b''):
            hash_obj.update(chunk)
    return hash_obj.hexdigest()
 
# Sniff packets pada antar muka jaringan
scapy.sniff(iface="eth0", prn=sniff_packets, filter="tcp")
 
# Jalankan Perintah lanjutan
output = run_cmd("netstat -ano")
print(output)
 
# Hash File
file_path = "/path/to/file"
file_hash = hash_file(file_path)
print(f"File Hash: {file_hash}")
```

**Dalam contoh ini, kami menambahkan dua fungsi tambahan:**

- 1.**run_cmd()** fungsi: Fungsi ini memungkinkan menjalankan perintah lanjutan menggunakan modul. Anda dapat memberikan perintah apa pun sebagai string, dan itu akan dieksekusi di shell.
- subprocess

2.**hash_file()** fungsi: Fungsi ini menghitung hash SHA-256 dari file menggunakan modul. Berikan jalur file sebagai input, dan itu akan mengembalikan nilai hash.
- hashlib
