##### Skrip Bash Ultra-Canggih:

```python
#!/bin/bash
 
# Berfungsi untuk memeriksa integritas file menggunakan SHA-256
check_file_integrity() {
  file="$1"
  expected_hash="$2"
  
  actual_hash=$(sha256sum "$file" | awk '{print $1}')
  
  if [[ "$actual_hash" == "$expected_hash" ]]; then
    echo "File: $file | Integrity check: Passed"
  else
    echo "File: $file | Integrity check: Failed"
  fi
}
 
# Berfungsi untuk melakukan DNS lookup dan reverse DNS lookup
dns_lookup() {
  domain="$1"
  
  ip_address=$(dig +short "$domain")
  
  if [[ -n "$ip_address" ]]; then
    echo "Domain: $domain | Alamat IP: $ip_address"
    
    reverse_domain=$(dig +noall +answer -x "$ip_address" | awk '{print $NF}')
    
    if [[ -n "$reverse_domain" ]]; then
      echo "Alamat IP: $ip_address | Reverse Domain: $reverse_domain"
    else
      echo "Alamat IP: $ip_address | Reverse Domain: Not found"
    fi
  else
    echo "Domain: $domain | Alamat IP: Tidak Ditemukan"
  fi
}
```

```python
import os
import hashlib
import socket
import requests
 
# Berfungsi untuk memeriksa integritas file menggunakan SHA-256

def check_file_integrity(file, expected_hash):
    with open(file, "rb") as f:
        content = f.read()
        actual_hash = hashlib.sha256(content).hexdigest()
        
        if actual_hash == expected_hash:
            print(f"File: {file} | Pemeriksaan integritas: Lulus")
        else:
            print(f"File: {file} | Pemeriksaan Integritas: Gagal")
 
#  Berfungsi untuk melakukan DNS lookup dan reverse DNS lookup

def dns_lookup(domain):
    try:
        ip_address = socket.gethostbyname(domain)
        print(f"Domain: {domain} | Alamat IP: {ip_address}")
        
        try:
            reverse_domain = socket.gethostbyaddr(ip_address)[0]
            print(f"Alamat IP: {ip_address} | Reverse Domain: {reverse_domain}")
        except socket.herror:
            print(f"Alamat IP: {ip_address} | Reverse Domain: Tidak Ditemukan")
    except socket.gaierror:
        print(f"Domain: {domain} | Alamat IP: Tidak Ditemukan")
```

```python
#!/bin/bash
 
# Berfungsi untuk melakukan DNS lookup dan reverse DNS lookup

check_file_integrity() {
  file="$1"
  expected_hash="$2"
  
  actual_hash=$(sha256sum "$file" | awk '{print $1}')
  
  if [[ "$actual_hash" == "$expected_hash" ]]; then
    echo "File: $file | Pemeriksaan Integritas: Lulus"
  else
    echo "File: $file | Pemeriksaan Integritas: Gagal"
  fi
}
 
# Berfungsi untuk melakukan DNS lookup dan reverse DNS lookup
dns_lookup() {
  domain="$1"
  
  ip_address=$(dig +short "$domain")
  
  if [[ -n "$ip_address" ]]; then
    echo "Domain: $domain | Alamat IP: $ip_address"
    
    reverse_domain=$(dig +noall +answer -x "$ip_address" | awk '{print $NF}')
    
    if [[ -n "$reverse_domain" ]]; then
      echo "Alamat IP: $ip_address | Reverse Domain: $reverse_domain"
    else
      echo "Alamat IP: $ip_address | Reverse Domain: Tidak Ditemukan"
    fi
  else
    echo "Domain: $domain | IP Address: Tidak ditemukan"
  fi
}
 
# Berfungsi untuk memeriksa kedaluwarsa sertifikat SSL / TLS
check_certificate_expiration() {
  domain="$1"
  
  expiration_date=$(openssl s_client -servername "$domain" -connect "$domain":443 2>/dev/null | openssl x509 -noout -dates | grep "notAfter" | awk -F "=" '{print $2}')
  
  if [[ -n "$expiration_date" ]]; then
    expiration_timestamp=$(date -d "$expiration_date" +%s)
    current_timestamp=$(date +%s)
    
    if [[ "$expiration_timestamp" -gt "$current_timestamp" ]]; then
      echo "Domain: $domain | Certificate Kadaluwarsa: $expiration_date | Status: Valid"
    else
      echo "Domain: $domain | Certificate Kadaluwarsa: $expiration_date | Status: Kadaluwarsa"
    fi
  else
    echo "Domain: $domain | Certificate Kadaluwarsa: Tidak Ditemukan | Status: Tidak Diketahui"
  fi
}
```

##### Skrip python ultra canggih:

```python
import os
import hashlib
import socket
import requests
import ssl
 
# Berfungsi untuk memeriksa integritas file menggunakan SHA-256

def check_file_integrity(file, expected_hash):
    with open(file, "rb") as f:
        content = f.read()
        actual_hash = hashlib.sha256(content).hexdigest()
        
        if actual_hash == expected_hash:
            print(f"File: {file} | Pemeriksaan Integritas: Lulus")
        else:
            print(f"File: {file} | Pemeriksaan Integritas: Gagal")
 
# Berfungsi untuk melakukan DNS lookup dan reverse DNS lookup
def dns_lookup(domain):
    try:
        ip_address = socket.gethostbyname(domain)
        print(f"Domain: {domain} | Alamat IP: {ip_address}")
        
        try:
            reverse_domain = socket.gethostbyaddr(ip_address)[0]
            print(f"Alamat IP: {ip_address} | Reverse Domain: {reverse_domain}")
        except socket.herror:
            print(f"Alamat IP: {ip_address} | Reverse Domain: Tidak Ditemukan")
    except socket.gaierror:
        print(f"Domain: {domain} | Alamat IP: Tidak Ditemukan")
 
# Berfungsi untuk memeriksa kedaluwarsa sertifikat SSL / TLS

def check_certificate_expiration(domain):
    context = ssl.create_default_context()
    
    try:
        with socket.create_connection((domain, 443)) as sock:
            with context.wrap_socket(sock, server_hostname=domain) as ssl_sock:
                cert = ssl_sock.getpeercert()
                expiration_date = cert['notAfter']
                current_date = ssl._ssl._test_decode_time(time.time())
                
                if expiration_date > current_date:
                    print(f"Domain: {domain} | Certificate Kadaluwarsa: {expiration_date} | Status: Valid")
                else:
                    print(f"Domain: {domain} | Certificate Kadaluwarsa: {expiration_date} | Status: Kadaluwarsa")
    except (socket.gaierror, socket.timeout, ConnectionRefusedError):
        print(f"Domain: {domain} | Certificate Kadaluwarsa: Tidak ditemukan | Status: Tidak diketahui")
```

**COntoh Penggunaan:**

```python
check_file_integrity("/path/to/file.txt", "5d41402abc4b2a76b9719d911017c592")
dns_lookup("example.com")
check_certificate_expiration("example.com")
```
