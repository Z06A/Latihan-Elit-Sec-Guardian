#### Berikut adalah contoh kode solusi untuk menonaktifkan layanan yang tidak perlu di Linux menggunakan skrip Bash:

```python
#!/bin/bash
 
# Daftar semua layanan yang berjalan pada sistem
services=$(systemctl list-units --type=service --state=running --no-pager | awk '{print $1}')
 
# Berbagai layanan yang akan dinonaktifkan
services_to_disable=("telnet" "ftp" "rsh" "rlogin" "rexec" "talk" "ntalk" "samba" "vsftpd")
 
# Loop melalui layanan dan menonaktifkannya
for service in "${services_to_disable[@]}"; do
  if [[ " ${services[@]} " =~ " ${service} " ]]; then
    systemctl disable $service
    systemctl stop $service
    echo "Layanan Dinonaktifkan dan dihentikan: $service"
  fi
done
 
echo "Penonaktifan Layanan selesai."
```

Skrip ini mencantumkan semua layanan yang berjalan dan memeriksa apakah ada layanan dalam array yang berjalan. Jika layanan ditemukan, itu dinonaktifkan dan berhenti menggunakan perintah. Skrip memberikan umpan balik yang jelas tentang layanan mana yang dinonaktifkan dan dihentikan.
- services_to_disable
- systemctl

Ingat, penting untuk menyesuaikan skrip dengan sistem operasi dan versi tertentu yang digunakan. Selain itu, peserta didik harus menguji skrip dalam lingkungan yang terkendali sebelum menerapkannya pada sistem produksi.

***Contoh Kode Bash Scripting:***

- Berikut adalah contoh skrip Bash yang menggabungkan beberapa teknik pengerasan OS tingkat lanjut:

```python
#!/bin/bash
 
# Perbarui dan tingkatkan sistem
apt update
apt upgrade -y
 
# Nonaktifkan layanan yang tidak perlu
services_to_disable=("telnet" "ftp" "rsh" "rlogin" "rexec" "talk" "ntalk" "samba" "vsftpd")
 
for service in "${services_to_disable[@]}"; do
  systemctl disable $service
  systemctl stop $service
done
 
# Konfigurasikan aturan firewall
firewall-cmd --zone=public --add-port=22/tcp --permanent
firewall-cmd --zone=public --add-service=http --permanent
firewall-cmd --reload
 
# Aktifkan pembaruan keamanan otomatis
apt install unattended-upgrades -y
echo 'APT::Periodic::Update-Package-Lists "1";' >> /etc/apt/apt.conf.d/20auto-upgrades
echo 'APT::Periodic::Unattended-Upgrade "1";' >> /etc/apt/apt.conf.d/20auto-upgrades
 
# Mengeraskan akun pengguna
sed -i 's/^PermitRootLogin yes/PermitRootLogin no/' /etc/ssh/sshd_config
sed -i 's/^PasswordAuthentication yes/PasswordAuthentication no/' /etc/ssh/sshd_config
systemctl restart sshd
 
# Mengeraskan akun pengguna
usermod -s /sbin/nologin root
useradd -M -s /sbin/nologin -c "Service Account" serviceuser
 
# Instal dan konfigurasikan sistem deteksi intrusi (IDS)
echo "fs.suid_dumpable = 0" >> /etc/sysctl.conf
echo "kernel.dmesg_restrict = 1" >> /etc/sysctl.conf
sysctl -p
 
# Install and configure intrusion detection system (IDS)
apt install snort -y
sed -i 's/^# frag3_global:/frag3_global:/' /etc/snort/snort.conf
sed -i 's/^# preprocessor sfportscan:/preprocessor sfportscan:/' /etc/snort/snort.conf
systemctl enable --now snort
 
echo "Pengerasan sistem operasi selesai."
```

Harap dicatat bahwa skrip ini memberikan gambaran umum tentang teknik pengerasan OS tingkat lanjut dan mungkin memerlukan penyesuaian berdasarkan persyaratan dan sistem operasi spesifik Anda. Selain itu, sangat penting untuk menguji skrip dalam lingkungan yang terkendali sebelum menerapkannya ke sistem produksi.

***Skrip Python (Pemeriksa Keamanan Siber Tingkat Lanjut):***

```python
import requests
import socket
import subprocess
import os
 
# Periksa apakah situs web aktif dan berjalan
def check_website(url):
    try:
        response = requests.get(url)
        if response.status_code == 200:
            print(f"The website {url} Aktif Dan Berjalan.")
        else:
            print(f"The website {url} Tidak Dapat Diakses.")
    except requests.exceptions.RequestException:
        print(f"Terjadi kesalahan saat mengakses situs web {url}.")
 
# Periksa apakah port tertentu terbuka pada alamat IP target
def check_port(ip, port):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    result = sock.connect_ex((ip, port))
    if result == 0:
        print(f"The port {port} terbuka pada alamat IP {ip}.")
    else:
        print(f"The port {port} ditutup pada alamat IP {ip}.")
    sock.close()
 

# Jalankan perintah sistem dan dapatkan hasilnya
def run_command(command):
    try:
        output = subprocess.check_output(command, shell=True, universal_newlines=True)
        print(output)
    except subprocess.CalledProcessError as e:
        print(f"Kesalahan Perintah eksekusi: {e}")
 
# Periksa apakah ada file atau direktori tertentu
def check_file(file_path):
    if os.path.exists(file_path):
        print(f"The file Atau directory {file_path} ada.")
    else:
        print(f"The file Atau directory {file_path} Tidak Ada.")
 
# Lakukan pemeriksaan keamanan siber
def perform_cybersecurity_checks():
    check_website("https://example.com")
    check_port("127.0.0.1", 80)
    run_command("netstat -tuln")
    check_file("/var/log/syslog")

# Jalankan pemeriksaan keamanan siber
perform_cybersecurity_checks()
```

Skrip Python mendemonstrasikan fungsi untuk memeriksa ketersediaan situs web, menguji apakah port tertentu terbuka, menjalankan perintah sistem, dan memeriksa keberadaan file atau direktori. ***Skrip Python (Pemeriksa Keamanan Siber Tingkat Lanjut yang Diperluas):***

```python
import requests
import socket
import subprocess
import os
import hashlib
 
# Periksa apakah situs web aktif dan berjalan
def check_website(url):
    try:
        response = requests.get(url)
        if response.status_code == 200:
            print(f"The website {url} Aktif Dan Berjalan.")
        else:
            print(f"The website {url} Tidak Dapat Diakses.")
    except requests.exceptions.RequestException:
        print(f"Terjadi kesalahan saat mengakses situs web {url}.")
 
# Periksa apakah port tertentu terbuka pada alamat IP target
def check_port(ip, port):
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    result = sock.connect_ex((ip, port))
    if result == 0:
        print(f"The port {port} terbuka pada alamat IP {ip}.")
    else:
        print(f"The port {port} ditutup pada alamat IP {ip}.")
    sock.close()
 
# Jalankan perintah sistem dan dapatkan hasilnya
def run_command(command):
    try:
        output = subprocess.check_output(command, shell=True, universal_newlines=True)
        print(output)
    except subprocess.CalledProcessError as e:
        print(f"Kesalahan eksekusi Perintah: {e}")
 
# Periksa apakah ada file atau direktori tertentu
def check_file(file_path):
    if os.path.exists(file_path):
        print(f"The file atau directory {file_path} Ada.")
    else:
        print(f"The file atau directory {file_path} Tidak Ada.")
 
# Hitung nilai hash file
def calculate_file_hash(file_path):
    try:
        with open(file_path, "rb") as file:
            data = file.read()
            hash_value = hashlib.sha256(data).hexdigest()
            print(f"The SHA256 hash of {file_path} Adalah: {hash_value}")
    except FileNotFoundError:
        print(f"File {file_path} Tidak Ada.")
 
# Lakukan pemeriksaan keamanan siber tingkat lanjut
def perform_advanced_cybersecurity_checks():
    perform_cybersecurity_checks()
 
    # Pemeriksaan Tambahan
    check_website("https://example2.com")
    check_port("127.0.0.1", 443)
    run_command("netstat -ant")
    check_file("/var/log/auth.log")
    calculate_file_hash("/etc/passwd")
 
# Jalankan pemeriksaan keamanan siber tingkat lanjut

perform_advanced_cybersecurity_checks()
```

Skrip Python yang diperluas ini mencakup pemeriksaan keamanan siber tambahan, seperti memeriksa situs web kedua, menguji port yang berbeda, menjalankan perintah netstat dengan opsi yang berbeda, memeriksa file log yang berbeda, dan menghitung hash SHA256 dari sebuah file.
