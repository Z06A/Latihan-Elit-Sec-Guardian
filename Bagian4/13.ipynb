Ethical Hacking menggunakan Kali Linux:
Kali Linux adalah distribusi populer untuk peretasan etis dan pengujian penetrasi. Muncul dengan berbagai alat pra-instal yang dapat digunakan untuk berbagai teknik peretasan. Berikut adalah contoh perintah untuk pemindaian jaringan menggunakan alat Nmap di Kali Linux:

***1.Pemindaian Jaringan dengan Nmap:***

- Instalasi: Nmap sudah diinstal sebelumnya di Kali Linux. Jika Anda perlu menginstalnya secara terpisah, gunakan perintah berikut:

```python
sudo apt-get install nmap
```

- Penggunaan: Untuk melakukan pemindaian jaringan dasar, gunakan perintah berikut:


```python
sudo nmap -sS <target_IP>
```

- Ganti dengan alamat IP atau rentang IP yang ingin Anda pindai.<target_IP>

*Contoh: Memindai satu IP target:*

```python
sudo nmap -sS 192.168.0.1
```

*Contoh: Memindai rentang IP:*

```python
sudo nmap -sS 192.168.0.1-20
```

Ingat, peretasan etis hanya boleh dilakukan dengan otorisasi yang tepat dan untuk tujuan yang sah.
