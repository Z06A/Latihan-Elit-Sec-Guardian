```python
Import Hashlib
 
# Fungsi untuk menghitung hash MD5 dari sebuah string
def calculate_md5_hash(string):
    md5_hash = hashlib.md5(string.encode()).hexdigest()
    return md5_hash
 
# Program utama
input_string = input("Masukkan string untuk menghitung hash MD5-nya: ")
md5_hash = calculate_md5_hash(input_string)
print("MD5 Hash:", md5_hash)
```

Untuk skrip Python tingkat lanjut, berikut adalah contoh yang menghasilkan kata sandi acak dan mengenkripsinya menggunakan pustaka bcrypt:

```python
import bcrypt
import random
import string
 
# Berfungsi untuk menghasilkan kata sandi acak
def generate_random_password(length):
    characters = string.ascii_letters + string.digits + string.punctuation
    password = ''.join(random.choice(characters) for _ in range(length))
    return password
 
# Berfungsi untuk mengenkripsi password menggunakan bcrypt
def encrypt_password(password):
    salt = bcrypt.gensalt()
    hashed_password = bcrypt.hashpw(password.encode(), salt)
    return hashed_password
 
# Program utama
password_length = 12
random_password = generate_random_password(password_length)
encrypted_password = encrypt_password(random_password)
 
print("Pasword yang di buat secara acak:", random_password)
print("Enkripsi password:", encrypted_password)
```
