**SSL / TLS:**
Secure Sockets Layer (SSL) dan penggantinya, Transport Layer Security (TLS), adalah protokol kriptografi yang menyediakan komunikasi aman melalui web. Kode solusi menunjukkan cara mengonfigurasi sertifikat SSL/TLS untuk server web dan mengaktifkan koneksi terenkripsi antara klien dan server.

**SSH:**
Secure Shell (SSH) adalah protokol kriptografi yang digunakan untuk akses jarak jauh yang aman ke sistem. Kode solusi menunjukkan cara menyiapkan autentikasi berbasis kunci SSH dan membatasi akses hanya untuk pengguna yang berwenang.

SSL / TLS adalah protokol kriptografi yang banyak digunakan untuk komunikasi yang aman melalui web. Di Kali Linux, Anda dapat menggunakan pustaka OpenSSL dan alat baris perintah untuk menghasilkan sertifikat SSL/TLS dan mengonfigurasi koneksi aman. 

#### Berikut adalah contoh bagaimana Anda dapat membuat sertifikat SSL/TLS yang ditandatangani sendiri menggunakan OpenSSL:

*Instal OpenSSL (jika belum diinstal):*


```python
sudo apt update
sudo apt install openssl
```

*Buat kunci privat dan sertifikat yang ditandatangani sendiri:*

```python
openssl req -x509 -nodes -days 365 -newkey rsa:2048 -keyout server.key -out server.crt
```

Berikan detail yang diperlukan saat diminta (seperti negara, organisasi, dll.). Anda dapat membiarkan bidang "Nama Umum" kosong atau memasukkan nama domain atau alamat IP server.

Kunci privat akan disimpan dalam file, dan sertifikat yang ditandatangani sendiri akan disimpan dalam file.server.key,server.crt

***Sekarang, mari kita beralih ke protokol IPsec:***

Di Kali Linux, Anda dapat menggunakan paket untuk mengonfigurasi terowongan IPsec. 

*Instal menggunakan perintah berikut:strongswan*

```python
sudo apt update
sudo apt install strongswan
```

Konfigurasikan IPsec dengan mengedit file. 

*Berikut adalah contoh konfigurasi untuk terowongan VPN situs-ke-situs:/etc/ipsec.conf*

```python
conn vpn-tunnel
    left=192.168.1.1
    right=192.168.2.1
    authby=secret
    auto=start
    type=tunnel
    ike=aes256-sha1-modp1024
    esp=aes256-sha1
```

- Ganti dan dengan alamat IP dari dua gateway jaringan.192.168.1.1,192.168.2.1

- Tambahkan kunci yang dibagikan sebelumnya dengan mengedit file:/etc/ipsec.secrets

```python
192.168.1.1 192.168.2.1 : PSK "Kunci yang dibagikan sebelumnya"
```

Ganti dengan kunci yang kuat yang dibagikan sebelumnya.your-pre-shared-key

**Mulai layanan IPsec:**

```python
sudo systemctl start strongswan
```

#### Sekarang, mari kita beralih ke protokol SSH:

SSH biasanya sudah diinstal sebelumnya di Kali Linux. Untuk mengaktifkan server SSH, gunakan perintah berikut:

```python
sudo systemctl enable ssh
sudo systemctl start ssh
```

Untuk meningkatkan keamanan, disarankan untuk menonaktifkan autentikasi berbasis kata sandi dan menggunakan autentikasi berbasis kunci SSH sebagai gantinya. Buat pasangan kunci SSH menggunakan perintah berikut:

```python
ssh-keygen -t rsa -b 4096
```

Salin kunci publik ke server jarak jauh yang ingin Anda sambungkan:

```python
ssh-copy-id username@remote-server
```

Ganti dengan nama pengguna Anda dan dengan alamat IP atau nama domain server jarak jauh.username,remote-server

#### Sekarang, Anda dapat terhubung ke server jarak jauh menggunakan autentikasi berbasis kunci SSH:

```python
ssh username@remote-server
```
