#### Sekarang mari kita beralih ke contoh kode lanjutan untuk analisis log di Bash.

**Skrip Bash Tingkat Lanjut untuk Analisis Log:**

```python
#!/bin/bash
 
log_file="system.log"
output_file="analysis_report.txt"
 
# Berfungsi untuk menganalisis file log
analyze_log() {
     # Cari pola tertentu dalam file log
    error_lines=$(grep "error" "$log_file")
    warning_lines=$(grep "warning" "$log_file")
 
    # Menghasilkan laporan analisis
    echo "Laporan Analysis Log" > "$output_file"
    echo "===================" >> "$output_file"
    echo "Errors:" >> "$output_file"
    echo "$error_lines" >> "$output_file"
    echo "" >> "$output_file"
    echo "Warnings:" >> "$output_file"
    echo "$warning_lines" >> "$output_file"
    echo "" >> "$output_file"
    echo "Analisis berhasil diselesaikan. Periksa $output_file untuk hasilnya."
}
 
# Jalankan Fungsi Analysis Log

analyze_log
```

*Penjelasan:*Skrip Bash di atas melakukan analisis log dengan mencari pola tertentu (misalnya, "kesalahan" dan "peringatan") dalam file log yang ditentukan. Kemudian menghasilkan laporan analisis dalam file output terpisah.

**Skrip Python Tingkat Lanjut untuk Keamanan Siber:**

```python
import os
import subprocess
 
# Lakukan Pemeriksaan keamanan
print("Memulai Melakukan Pemeriksaan Keamanan Cyber..")
 
# Periksa Konektivitas jaringan
ping_result = subprocess.run(["ping", "-c", "4", "google.com"], stdout=subprocess.DEVNULL).returncode
if ping_result == 0:
    print("Koneksi Jaringan: OKE")
else:
    print("Koneksi Jaringan: Gagal")
 
# Periksa Port Terbuka
open_ports = subprocess.check_output("nmap -p- localhost | grep 'open' | awk -F/ '{print $1}'", shell=True).decode().strip()
print("Open ports:", open_ports)
 
# Periksa Kerentanan System
vulnerabilities = subprocess.check_output("sudo openvas --update && sudo openvas --start && sudo openvas --get-tasks && sudo openvas --get-results", shell=True).decode().strip()
print("kerentanan System:", vulnerabilities)
 
# Lakukan Pemindaian Malware
malware_scan = subprocess.check_output("sudo clamscan -r /path/to/scan | grep 'Infected files'", shell=True).decode().strip()
print("Pemindaian Malware:", malware_scan)
 
print("Cybersecurity checks completed.")
```

**Skrip Bash Tingkat Lanjut untuk Keamanan Siber:**

```python
#!/bin/bash
 
# Lakukan Pemeriksaan Keamanan Cyeber
echo "Memulai Memeriksa Keamanan Cyber.."
 
# Periksa koneksi jaringan 
ping -c 4 google.com > /dev/null
if [ $? -eq 0 ]; then
    echo "Koneksi Jaringan: OKE"
else
    echo "Koneksi Jaringan: Gagal"
fi
 
# Periksa Port terbuka
open_ports=$(sudo nmap -p- localhost | grep -E 'open|filtered' | awk -F/ '{print $1}' | tr '\n' ', ')
echo "Buka  ports: $open_ports"
 
# Pemeriksaan Kerentanan System
(vuln_status=$(sudo openvas --update && sudo openvas --start && sudo openvas --get-tasks && sudo openvas --get-results) && echo "Kerentanan System: $vuln_status") || echo "Kerentanan sistem: Terjadi kesalahan saat memeriksa kerentanan"
 
# Lakukan Pemindaian Malware
malware_scan=$(sudo clamscan -r /path/to/scan --exclude-dir=/path/to/exclude | grep 'Infected files')
echo "Pemindaian Malware: $malware_scan"
 
# Periksa Status Firewal
firewall_status=$(sudo ufw status | grep 'Status' | awk '{print $2}')
echo "Status Firewall: $firewall_status"
 
# Periksa Pengguna disk
disk_usage=$(df -h | awk '/^\/dev/{print $1,$5}')
echo "Pengguna DIsk:"
echo "$disk_usage"
 
echo "pemeriksa Keamanan Cyber selesai."
```

Penjelasan: Skrip Bash yang disempurnakan mencakup fitur tambahan seperti memeriksa port yang terbuka, mengecualikan direktori tertentu dari pemindaian malware, memeriksa status firewall menggunakan UFW, dan menampilkan informasi penggunaan disk menggunakan perintah.df

**Skrip Python Tingkat Lanjut untuk Keamanan Siber:**

```python
import os
import subprocess
 
Lakukan Pemeriksaan Keamanan Cyber
print("Memulai Memeriksa Keamanan Cyber..")
 
# Peeriksaan Koneksi Jaringan
ping_result = subprocess.run(["ping", "-c", "4", "google.com"], stdout=subprocess.DEVNULL).returncode
if ping_result == 0:
    print("Koneksi Jaringan: OKE")
else:
    print("Koneksi Jaringan: Gagal")
 
# Pemeriksaan port yang terbuka
open_ports = subprocess.check_output("sudo nmap -p- localhost | grep -E 'open|filtered' | awk -F/ '{print $1}' | tr '\n' ', '", shell=True).decode().strip()
print("Buka Port:", open_ports)
 
# Pengecekan Kerentanan System
try:
    vulnerabilities = subprocess.check_output("sudo openvas --update && sudo openvas --start && sudo openvas --get-tasks && sudo openvas --get-results", shell=True, stderr=subprocess.STDOUT).decode().strip()
    print("Kerentanan System :", vulnerabilities)
except subprocess.CalledProcessError as e:
    print("Kerentanan sistem: Terjadi kesalahan saat memeriksa kerentanan")
 
# Lakukan Pemindaian Malware
malware_scan = subprocess.check_output("sudo clamscan -r /path/to/scan --exclude-dir=/path/to/exclude | grep 'Infected files'", shell=True).decode().strip()
print("Pemindaian Malware:", malware_scan)
 
# Periksa Status Firewall
firewall_status = subprocess.check_output("sudo ufw status | grep 'Status' | awk '{print $2}'", shell=True).decode().strip()
print("Status Firewall:", firewall_status)
 
# Periksa Penggunna disk
disk_usage = subprocess.check_output("df -h | awk '/^\/dev/{print $1,$5}'", shell=True).decode().strip()
print("Pengguna Disk:")
print(disk_usage)
 
print("Pemeriksaan Keamanan Cyber Selesai.")
```

*Penjelasan:* Skrip Python tingkat lanjut mencakup fitur tambahan yang mirip dengan skrip Bash yang ditingkatkan. Ini menggunakan modul untuk menjalankan perintah shell dan menangkap outputnya.subprocess
