*Contoh berikut menunjukkan cara melakukan pemindaian port menggunakan pustaka dengan Python:nmap*

```python
import nmap
 
def port_scan(target_host, target_ports):
    nm = nmap.PortScanner()
    scan_result = nm.scan(target_host, target_ports)
    
    for host in scan_result['scan']:
        print(f"Buka Port Untuk {host}:")
        for port in scan_result['scan'][host]['tcp']:
            if scan_result['scan'][host]['tcp'][port]['state'] == 'open':
                print(f"Port {port}/tcp is open")
```

**Contoh Penggunnaan:**

```python
target_host = '127.0.0.1'
target_ports = '1-1000'
port_scan(target_host, target_ports)
```

**Catatan:** Sebelum menjalankan skrip ini, pastikan Anda telah menginstal pustaka. Anda dapat menginstalnya menggunakan file .nmap,pip install python-nmap

Pindah ke contoh skrip Bash, berikut adalah skrip lanjutan yang melakukan pemindaian jaringan sederhana menggunakan perintah:ping


```python
#!/bin/bash
 
network="192.168.1"
 
for host in $(seq 1 254); do
    ip_address="$network.$host"
    ping -c 1 -W 1 "$ip_address" > /dev/null 2>&1
    if [ $? -eq 0 ]; then
        echo "Host $ip_address Sudah Habis"
    fi
done
```

Simpan kode di atas dalam file (misalnya, ), membuatnya dapat dieksekusi (), dan jalankan menggunakan .network_scan.sh,chmod +x network_scan.sh./network_scan.sh

**Untuk perintah CMD Windows, berikut adalah contoh pemindaian jaringan dasar menggunakan perintah:ping**

```python
@echo off
 
set network=192.168.1
 
for /L %%i in (1,1,254) do (
    set ip_address=%network%.%%i
    ping -n 1 -w 1000 %ip_address% | find "TTL=" > nul
    if not errorlevel 1 (
        echo Host %ip_address% Sudah Habis
    )
)
```

Simpan kode di atas dalam file (misalnya, ) dan jalankan dengan mengklik dua kali pada file tersebut.network_scan.bat
