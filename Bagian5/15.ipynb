#### Topik 1: Keamanan siber

Keamanan siber adalah bidang luas yang mencakup berbagai aspek melindungi sistem dan jaringan komputer dari akses, serangan, dan pelanggaran data yang tidak sah. 

*Berikut adalah contoh skrip Python yang menunjukkan pemindaian jaringan dasar menggunakan pustaka Scpy:*

```python
import scapy.all as scapy
 
def scan_network(target_ip):
    arp_request = scapy.ARP(pdst=target_ip)
    broadcast = scapy.Ether(dst="ff:ff:ff:ff:ff:ff")
    arp_request_broadcast = broadcast/arp_request
    answered_list = scapy.srp(arp_request_broadcast, timeout=1, verbose=False)[0]
 
    clients = []
    for element in answered_list:
        client_dict = {"ip": element[1].psrc, "mac": element[1].hwsrc}
        clients.append(client_dict)
    return clients
 
target_ip = "192.168.1.1/24"
scan_result = scan_network(target_ip)
 
for client in scan_result:
    print(f"IP: {client['ip']}\t MAC: {client['mac']}")
```

Skrip ini menggunakan pustaka Scapy untuk mengirim permintaan Address Resolution Protocol (ARP) untuk menemukan perangkat di jaringan.


#### Topik 2: Forensik Komputer

Forensik komputer melibatkan penyelidikan perangkat digital dan mengumpulkan bukti untuk tujuan hukum. 

*Berikut adalah contoh skrip Bash yang menggunakan alat baris perintah Sleuth Kit (TSK) untuk analisis disk dasar:*

```python
#!/bin/bash
 
image_path="/path/to/disk/image.dd"
 
# Analisis gambar disk
mmls "$image_path"  # Tampilkan Partisi disk
fsstat "$image_path"  # Menampilkan Informasi System File
 
# Pasang Gambar disk
mkdir /mnt/forensic
mount -o ro,loop "$image_path" /mnt/forensic
 
# Jelajahi System File
ls -l /mnt/forensic
cat /mnt/forensic/somefile.txt
 
# Lepaskan Gambar Disk
umount /mnt/forensic
```

Skrip ini menunjukkan cara menganalisis image disk, memasangnya hanya-baca, dan melakukan eksplorasi sistem file dasar menggunakan alat TSK seperti 
- , , , dan .
