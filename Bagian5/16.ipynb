**1.Kriptografi dan Komunikasi Aman dengan Python:**

Python adalah bahasa serbaguna untuk menerapkan kriptografi dan komunikasi yang aman.

*Berikut adalah contoh cuplikan kode yang mencakup berbagai aspek:*

```python
import cryptography
from cryptography.hazmat.primitives import hashes, hmac
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
 
# Enkripsi Simetris (AES)
key = b'my_secret_key'
cipher = Cipher(algorithms.AES(key), modes.CBC())
encryptor = cipher.encryptor()
ciphertext = encryptor.update(b'my_message') + encryptor.finalize()
 
# Dekripsi Simetris (AES)
decryptor = cipher.decryptor()
decrypted_message = decryptor.update(ciphertext) + decryptor.finalize()
 
# Enkripsi Asimetris (RSA)
private_key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
public_key = private_key.public_key()
ciphertext = public_key.encrypt(b'my_message', padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()), algorithm=hashes.SHA256(), label=None))
 
# Dekripsi Asimetris (RSA)
decrypted_message = private_key.decrypt(ciphertext, padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()), algorithm=hashes.SHA256(), label=None))
 
# Tanda Tangan Digital / Digital Signature(RSA)

message = b'my_message'
signature = private_key.sign(message, padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH), hashes.SHA256())
 
# Verifikasi Tanda Tangan/ signature verivication (RSA)

public_key.verify(signature, message, padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH), hashes.SHA256())
```

**2.Kriptografi dan Komunikasi Aman di Bash Scripting:**

Bash scripting tidak umum digunakan untuk operasi kriptografi tingkat rendah, tetapi Anda dapat menggunakan alat baris perintah yang ada untuk enkripsi, dekripsi, dan tanda tangan digital. 

*Berikut adalah contoh menggunakan OpenSSL untuk enkripsi dan dekripsi:*

```python
# Enkripsi Simetris (AES)

echo -n "my_message" | openssl enc -e -aes-256-cbc -K 6b6579636f64652d6b6579 -iv 6b6579636f64652d6b6579 -base64
 
# Dekripsi Simetris (AES)

echo -n "encrypted_base64_message" | openssl enc -d -aes-256-cbc -K 6b6579636f64652d6b6579 -iv 6b6579636f64652d6b6579 -base64
 
# Enkripsi Asimetris (RSA)

echo -n "my_message" | openssl rsautl -encrypt -pubin -inkey public_key.pem -out encrypted_message.rsa
 
# Dekripsi Asimetris (RSA)

openssl rsautl -decrypt -inkey private_key.pem -in encrypted_message.rsa -out decrypted_message.txt
 
# Tanda Tangan Digital (RSA)

openssl dgst -sha256 -sign private_key.pem -out signature.bin my_message.txt
 
# Verifikasi Tanda Tangan (RSA)

openssl dgst -sha256 -verify public_key.pem -signature signature.bin my_message.txt
```
