##### Berikut adalah contoh cuplikan kode untuk menunjukkan teknik pemindaian port dasar menggunakan Python:

```python
import socket
 
def port_scan(target_host, target_ports):
    for port in target_ports:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(1)
        result = sock.connect_ex((target_host, port))
        if result == 0:
            print(f"Port {port} Teerbuka")
        sock.close()
 
target = "127.0.0.1"
ports = [21, 22, 80, 443, 8080]  # contoh Port Yang di Pindai
 
port_scan(target, ports)
```

***2.Cryptanalysis di Python:***

Cryptanalysis melibatkan analisis dan pemecahan sistem kriptografi. Berikut adalah contoh serangan analisis frekuensi sederhana pada sandi Caesar:

```python
def caesar_decrypt(ciphertext, shift):
    plaintext = ""
    for char in ciphertext:
        if char.isalpha():
            ascii_offset = 65 if char.isupper() else 97
            decrypted_char = chr((ord(char) - ascii_offset - shift) % 26 + ascii_offset)
            plaintext += decrypted_char
        else:
            plaintext += char
    return plaintext
 
ciphertext = "Bpqa qa i bmab!"
shift = 1
 
decrypted_text = caesar_decrypt(ciphertext, shift)
print(decrypted_text)
```


Dalam contoh ini, kita mendefinisikan fungsi yang mengambil ciphertext dan nilai shift sebagai input. Ini melakukan dekripsi Caesar dengan menggeser setiap karakter kembali dengan jumlah yang ditentukan. Teks yang didekripsi kemudian dicetak.

- caesar_decrypt

##### Sekarang, mari kita beralih ke kode yang disarankan untuk cryptanalysis tingkat lanjut dengan Python:

```python
Impor itertools
 
import itertools
 
def vigenere_decrypt(ciphertext, key):
    plaintext = ""
    key_length = len(key)
    key_indices = [ord(char.upper()) - 65 for char in key]
    for index, char in enumerate(ciphertext):
        if char.isalpha():
            ascii_offset = 65 if char.isupper() else 97
            key_index = index % key_length
            key_shift = key_indices[key_index]
            decrypted_char = chr((ord(char) - ascii_offset - key_shift) % 26 + ascii_offset)
            plaintext += decrypted_char
        else:
            plaintext += char
    return plaintext
 
ciphertext = "Tzxvvt, kgxw yivebv!"
possible_keys = ['KEY', 'SECRET', 'PASSWORD']
 
for key in possible_keys:
    decrypted_text = vigenere_decrypt(ciphertext, key)
    print(f"Key: {key}\nDecrypted Text: {decrypted_text}\n")
```

Dalam contoh ini, kami mendefinisikan fungsi yang melakukan dekripsi Vigenere menggunakan kunci yang diberikan. Fungsi ini mengambil ciphertext dan kunci sebagai input. Ini mengulangi setiap karakter dalam ciphertext, menerapkan algoritma dekripsi Vigenere menggunakan karakter kunci yang sesuai.
