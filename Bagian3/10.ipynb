#### 7.Berikut adalah contoh skrip Python yang menggabungkan teknik keamanan siber tingkat lanjut:


```python
import requests
from bs4 import BeautifulSoup
import hashlib
import subprocess
 
# Berfungsi untuk mengikis website dan mengekstrak link
def scrape_website(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    links = []
    for link in soup.find_all('a'):
        links.append(link.get('href'))
    return links
 
# Fungsi untuk hash password menggunakan SHA-256
def hash_password(password):
    salt = b'somerandomsalt'
    hashed_password = hashlib.pbkdf2_hmac('sha256', password.encode('utf-8'), salt, 100000)
    return hashed_password.hex()
 
# Berfungsi untuk melakukan pemindaian jaringan menggunakan Nmap
def perform_network_scan(target):
    result = subprocess.run(['nmap', '-p', '1-1000', '-T4', target], capture_output=True, text=True)
    return result.stdout
 
# Main function
def main():
    # Scrape a website and extract links
    website_links = scrape_website('http://example.com')
    print("Tauta ditemykan di situs web:")
    for link in website_links:
        print(link)
 
    # Hash kata sandi
    password = "mysecretpassword"
    hashed_password = hash_password(password)
    print("Hash password:", hashed_password)
 
    # Lakukan pemindaian jaringan
    target_ip = "192.168.1.1"
    scan_result = perform_network_scan(target_ip)
    print("Hasil pemindaian jaringan:")
    print(scan_result)
 
# Jalankan FUngsi Utama
if __name__ == "__main__":
    main()
```

### Berikut adalah versi yang disempurnakan dari skrip Python sebelumnya, menggabungkan fitur yang lebih canggih untuk tujuan keamanan siber:

```python
import requests
from bs4 import BeautifulSoup
import hashlib
import subprocess
import os
 
# Berfungsi untuk mengikis website dan mengekstrak link
def scrape_website(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    links = []
    for link in soup.find_all('a'):
        links.append(link.get('href'))
    return links
 

# Berfungsi untuk hash password menggunakan bcrypt
def hash_password(password):
    salt = bcrypt.gensalt()
    hashed_password = bcrypt.hashpw(password.encode('utf-8'), salt)
    return hashed_password.decode('utf-8')
 
# Berfungsi untuk melakukan pemindaian jaringan menggunakan Nmap
def perform_network_scan(target):
    result = subprocess.run(['nmap', '-p', '1-1000', '-T4', target], capture_output=True, text=True)
    return result.stdout
 
# Berfungsi untuk mengenkripsi file menggunakan GPG
def encrypt_file(file_path, recipient):
    encrypted_file_path = file_path + '.gpg'
    subprocess.run(['gpg', '--encrypt', '--recipient', recipient, '--output', encrypted_file_path, file_path])
    return encrypted_file_path
 
# Fungsi Utama
def main():
    # Mengikis Situs Web dan Mengekstrak Tautan
    website_links = scrape_website('http://example.com')
    print("Tautan Di temukan Di situs Web:")
    for link in website_links:
        print(link)
 
    # Hash Kata Sandi
    password = "mysecretpassword"
    hashed_password = hash_password(password)
    print("Katasandi Hash:", hashed_password)
 
    # Lakukan Peminndaian Jaringan
    target_ip = "192.168.1.1"
    scan_result = perform_network_scan(target_ip)
    print("Hasil Pemindaian Jaringan:")
    print(scan_result)
 
    # Enkripsi File
    file_path = 'secret.txt'
    recipient = 'user@example.com'
    encrypted_file_path = encrypt_file(file_path, recipient)
    print("File dienkripsi dan disimpan sebagai:", encrypted_file_path)
 
    # Delete File asli
    os.remove(file_path)
 
# Jalankan FUngsi Utama
if __name__ == "__main__":
    main()
```

#### Dalam versi lanjutan ini, saya menambahkan fitur-fitur berikut:

- 1.Hashing kata sandi menggunakan perpustakaan untuk keamanan yang lebih kuat.bcrypt
- 2.Enkripsi file menggunakan GPG (GNU Privacy Guard) untuk mengenkripsi file dan melindungi isinya.
- 3.Penghapusan file asli setelah enkripsi untuk keamanan data.
