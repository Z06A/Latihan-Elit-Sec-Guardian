Sekarang, mari selami lebih dalam topik TCP / IP. TCP / IP adalah protokol dasar yang digunakan untuk komunikasi di internet. Ini terdiri dari beberapa lapisan, masing-masing melayani tujuan tertentu. Memahami TCP / IP sangat penting untuk keamanan jaringan dan forensik komputer.

#### Model TCP / IP terdiri dari empat lapisan utama: lapisan aplikasi, lapisan transport, lapisan internet, dan lapisan tautan.

***1.Lapisan aplikasi:***

- Lapisan ini berkaitan dengan protokol dan layanan yang berinteraksi langsung dengan aplikasi, seperti HTTP, FTP, SMTP, dan DNS.

Tugas Pemrograman: Tulis skrip Python untuk mengirim permintaan HTTP ke server web dan mengambil respons.

***2.Lapisan transportasi:***

- Lapisan transport memastikan pengiriman data yang andal antar host. Ini menggunakan protokol seperti TCP (Transmission Control Protocol) dan UDP (User Datagram Protocol).

Tugas Pemrograman: Mengembangkan skrip Python untuk membuat koneksi TCP dengan server jarak jauh dan bertukar data.

***3.Lapisan internet:***

- Lapisan internet menangani pengalamatan IP, perutean, dan fragmentasi paket. Ini menggunakan IP (Internet Protocol) untuk pengalamatan dan perutean paket.

Tugas Pemrograman: Buat program Python untuk melakukan validasi alamat IP dan mengekstrak informasi jaringan dan host.

***4.Lapisan Tautan:***

- Lapisan link berhubungan dengan koneksi jaringan fisik, termasuk Ethernet, Wi-Fi, dan Bluetooth. Ini menangani framing dan transmisi data.

Tugas Pemrograman: Buat skrip Python untuk menangkap paket jaringan menggunakan antarmuka jaringan dan menganalisis header paket.

##### 2.Contoh kode lanjutan (Python): Koneksi soket TCP

```python
import socket
 
target_host = "www.example.com"
target_port = 80
 
# bUAT OBJECK SOCKET
client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
 
# Hubungkan ke host dan port target
client.connect((target_host, target_port))
 
# Kirim data keserver
client.send(b"GET / HTTP/1.1\r\nHost: www.example.com\r\n\r\n")
 
# terima data respons
response = client.recv(4096)
 
# Cetak data respons
print(response)
 
# tutup koneksi
client.close()
```

##### 3.Contoh Kode Tingkat Lanjut (Python): Validasi Alamat IP dan Ekstraksi Informasi

```python
import ipaddress
 
def validate_ip_address(ip):
    try:
        ipaddress.ip_address(ip)
        return True
    except ValueError:
        return False
 
def extract_ip_info(ip):
    ip_obj = ipaddress.ip_address(ip)
    network = ip_obj.network
    network_address = network.network_address
    broadcast_address = network.broadcast_address
    return {
        "Jaringan": str(network),
        "Alamat jaringan": str(network_address),
        "Alamat Siaran": str(broadcast_address),
        "Total Hosts": network.num_addresses - 2
    }
 
# Contoh Penggunaan
ip = "192.168.0.1"
if validate_ip_address(ip):
    ip_info = extract_ip_info(ip)
    for key, value in ip_info.items():
        print(f"{key}: {value}")
else:
    print("Alamat IP Tidak Valid.")
```

#### 7.Contoh Kode Keamanan Siber (Python): Pemeriksa Kekuatan Kata Sandi

```python
import re
 
def check_password_strength(password):
    length_error = len(password) < 8
    uppercase_error = re.search(r"[A-Z]", password) is None
    lowercase_error = re.search(r"[a-z]", password) is None
    digit_error = re.search(r"\d", password) is None
    symbol_error = re.search(r"\W", password) is None
 
    errors = []
    if length_error:
        errors.append("Panjang kata sandi minimal 8 karakter.")
    if uppercase_error:
        errors.append("Kata sandi harus berisi setidaknya satu huruf besar.")
    if lowercase_error:
        errors.append("Kata sandi harus berisi setidaknya satu huruf kecil")
    if digit_error:
        errors.append("Kata sandi harus berisi setidaknya satu digit.")
    if symbol_error:
        errors.append("Kata sandi harus berisi setidaknya satu karakter khusus.")
 
    if errors:
        return False, errors
    else:
        return True, []
```

***Contoh Penggunaan:***

```python
password = "MySecurePassword123!"
is_strong, error_messages = check_password_strength(password)
if is_strong:
    print("Kata Sandi Kuat")
else:
    print("Kata sandi Lemah. Error(s):")
    for error in error_messages:
        print(f"- {error}")
```

#### 8.Contoh skrip bash: pemeriksa integritas file

***Berikut kelanjutan dari Contoh Bash Script:***

```python
#!/bin/bash
 
# Hitung hash MD5 dari sebuah file
function calculate_md5() {
  md5sum "$1" | awk '{ print $1 }'
}
 
# Verifikasi integritas file
function verify_integrity() {
  file="$1"
  expected_hash="$2"
 
  if [[ -f "$file" ]]; then
    actual_hash=$(calculate_md5 "$file")
    if [[ "$actual_hash" == "$expected_hash" ]]; then
      echo "Pemeriksaan integritas lulus: $file"
    else
      echo "Pemeriksaan integritas gagal: $file"
    fi
  else
    echo "File tidak ditemukan: $file"
  fi
}
```

***Contoh Penggunaan: ***

```python
file="example.txt"
expected_hash="5eb63bbbe01eeed093cb22bb8f5acdc3"
verify_integrity "$file" "$expected_hash"
```

#### Berikut adalah contoh skrip Bash yang menggabungkan elemen pemindaian jaringan dan pemeriksaan kekuatan kata sandi:

```python
#!/bin/bash
 
# Pemindaian jaringan dengan Nmap
function network_scan() {
  target="$1"
  nmap -A -T4 "$target"
}
 
# Pemeriksaan kekuatan kata sandi
function check_password_strength() {
  password="$1"
  length_error=$(expr length "$password" '<' 8)
  uppercase_error=$(expr "$password" : '.*[A-Z]' '=' 0)
  lowercase_error=$(expr "$password" : '.*[a-z]' '=' 0)
  digit_error=$(expr "$password" : '.*[0-9]' '=' 0)
  symbol_error=$(expr "$password" : '.*[^[:alnum:]]' '=' 0)
 
  errors=()
  if [[ $length_error -eq 1 ]]; then
    errors+=("Panjang kata sandi minimal harus 8 karakter.")
  fi
  if [[ $uppercase_error -eq 1 ]]; then
    errors+=("Kata sandi harus berisi setidaknya satu huruf besar.")
  fi
  if [[ $lowercase_error -eq 1 ]]; then
    errors+=("Kata sandi harus berisi setidaknya satu huruf keci.")
  fi
  if [[ $digit_error -eq 1 ]]; then
    errors+=("Kata sandi harus berisi setidaknya satu digit.")
  fi
  if [[ $symbol_error -eq 1 ]]; then
    errors+=("Kata sandi harus berisi setidaknya satu karakter khusus.")
  fi
 
  if [[ ${#errors[@]} -eq 0 ]]; then
    echo "Kata sandi kuat."
  else
    echo "Kata sandi lemah. Error(s):"
    for error in "${errors[@]}"; do
      echo "- $error"
    done
  fi
}
 
# Program Utama
network_scan "192.168.0.1"
check_password_strength "MySecurePassword123!"
```
