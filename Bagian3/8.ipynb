##### Contoh skrip bash (tingkat lanjut):

```python
#!/bin/bash
 
# Skrip Keamanan Siber Tingkat Lanjut
 
# Berfungsi untuk melakukan port scan menggunakan nmap
port_scan() {
  echo "Melakukan Pemindaian Port Menggunakan nmap..."
  nmap -p- -T4 -A -v $1
}
 
# Berfungsi untuk memeriksa header keamanan sebuah website menggunakan curl

check_security_headers() {
  echo "Memeriksa Header Keamanan $1..."
  curl -I $1 | grep -E 'Server:|X-Powered-By:|Strict-Transport-Security:|X-Content-Type-Options:|X-Frame-Options:|Content-Security-Policy:'
}
 
# Berfungsi untuk melakukan serangan brute-force password menggunakan Hydra
password_brute_force() {
  echo "Melakukan serangan brute-force kata sandi menggunakan Hydra ..."
  hydra -l $1 -P $2 $3 $4 $5
}
 
# Main menu
while true; do
  clear
  echo "===== Skrip Keamanan Siber Tingkat Lanjut ====="
  echo "1. Lakukan pemindaian port"
  echo "2. Periksa header keamanan situs web"
  echo "3. Lakukan serangan brute-force kata sandi"
  echo "4. Keluar"
  read -p "Masukkan pilihan Anda: " choice
 
  case $choice in
    1)
      read -p "Masukkan alamat IP target atau nama host: " target
      port_scan $target
      read -p "Tekan enter untuk melanjutkan..."
      ;;
    2)
      read -p "Masukkan URL situs web: " website
      check_security_headers $website
      read -p "Tekan enter untuk melanjutkan..."
      ;;
    3)
      read -p "Masukkan nama pengguna: " username
      read -p "Masukkan jalur file daftar kata sandi: " password_list
      read -p "Masukkan alamat IP target atau nama host: " target
      read -p "Masukkan protokol (http-get-form, http-post-form, dll.): " protocol
      read -p "Masukkan detail formulir login: " login_form
      password_brute_force $username $password_list $target $protocol $login_form
      read -p "Tekan enter untuk melanjutkan..."
      ;;
    4)
      echo "Keluar Dari script..."
      exit 0
      ;;
    *)
      echo "Pilihan Tidak Valid Silahkan Ulangin lagi."
      read -p "Tekan enter untuk melanjutkan..."
      ;;
  esac
done
```

##### Contoh skrip Python (tingkat lanjut):

```python
import subprocess
 
# Skrip Keamanan Siber Tingkat Lanjut
 
# Berfungsi untuk melakukan port scan menggunakan nmap
def port_scan(target):
    print("Melakukan pemindaian port menggunakan nmap...")
    subprocess.run(["nmap", "-p-", "-T4", "-A", "-v", target])
 
# Berfungsi untuk memeriksa header keamanan sebuah website menggunakan curl
def check_security_headers(website):
    print("Memeriksa header keamanan", website)
    subprocess.run(["curl", "-I", website])
 
# Berfungsi untuk melakukan serangan brute-force password menggunakan Hydra
def password_brute_force(username, password_list, target, protocol, login_form):
    print("Melakukan serangan brute-force kata sandi menggunakan Hydra...")
    subprocess.run(["hydra", "-l", username, "-P", password_list, target, protocol, login_form])
 
# Main menu
while True:
    print("===== Skrip Keamanan Siber Tingkat Lanjut =====")
    print("1. Lakukan pemindaian port")
    print("2.  Periksa header keamanan situs web")
    print("3. Lakukan serangan brute-force kata sandi")
    print("4. Keluar")
    choice = input("Masukan Pilihan Anda: ")
 
    if choice == "1":
        target = input("Masukkan alamat IP target atau nama host: ")
        port_scan(target)
        input("Tekan enter untuk melanjutkan...")
    elif choice == "2":
        website = input("Masukkan URL situs web: ")
        check_security_headers(website)
        input("Tekan enter untuk melanjutkan...")
    elif choice == "3":
        username = input("Masukkan nama pengguna: ")
        password_list = input("Enter the password list file path: ")
        target = input("Masukkan alamat IP target atau nama host: ")
        protocol = input("Masukkan protokol (http-get-form, http-post-form, dll.): ")
        login_form = input("Masukkan detail formulir login: ")
        password_brute_force(username, password_list, target, protocol, login_form)
        input
```
