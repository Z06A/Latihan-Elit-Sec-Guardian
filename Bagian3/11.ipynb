**2.Contoh Bash Scripting:**

Berikut adalah contoh skrip Bash yang menggunakan alat ini untuk memindai berbagai alamat IP dan menyimpan hasilnya ke file:
- nmap

```python
#!/bin/bash
 
# Tentukan rentang IP untuk dipindai
start_ip="192.168.0.1"
end_ip="192.168.0.100"
 
# Buat stempel waktu untuk file keluaran
timestamp=$(date +"%Y%m%d%H%M%S")
 
# Jalankan pemindaian nmap dan simpan hasilnya ke file
nmap -oN "scan_results_${timestamp}.txt" "$start_ip-$end_ip"
```

Skrip ini mendefinisikan alamat IP awal dan akhir, menghasilkan stempel waktu untuk file output, dan kemudian menjalankan perintah dengan rentang IP yang ditentukan. Hasil pemindaian disimpan ke file dengan stempel waktu dalam nama file.nmap

**3.Contoh Python Scripting:**

Berikut adalah contoh skrip Python yang menggunakan modul untuk melakukan pemindaian jaringan dan mencetak port terbuka dari alamat IP target:
- nmap

```python
import nmap
 
# Buat instance baru dari kelas PortScanner
scanner = nmap.PortScanner()
 
# Tentukan alamat IP target
target_ip = "192.168.0.1"
 
# Lakukan pemindaian pada IP target
scanner.scan(target_ip, arguments="-p1-1000")
 
# Cetak port yang terbuka
for host in scanner.all_hosts():
    print(f"Buka untuk {host}:")
    for port in scanner[host].all_tcp():
        if scanner[host].tcp(port).state() == "Buka":
            print(f"Port {port} Terbuka")
```

Skrip ini menggunakan modul dengan Python untuk membuat instance kelas. Kemudian melakukan pemindaian pada alamat IP target yang ditentukan dan mencetak port terbuka yang ditemukan.
- nmap
- PortScanner

```python
whois <target IP atau nama host>
```

- **Melakukan pencarian DNS:**

```python
nslookup <IP target atau nama host>
```
